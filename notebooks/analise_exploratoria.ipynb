## 1. Introdução ao problema
Este notebook tem como objetivo analisar detalhadamente a saúde pública nos municípios brasileiros entre os anos de 2015 e 2024, com o objetivo principal de identificar disparidades regionais e correlações críticas entre o acesso a recursos médicos, a eficácia das campanhas de imunização e os índices demográficos e epidemiológicos. A relevância desta pesquisa reside na demonstração de como a ciência de dados e a programação aplicada à saúde podem apoiar estrategicamente gestores públicos na tomada de decisões e na alocação eficiente de recursos hospitalares.

## 2. Explicação da base de dados
A base de dados utilizada (`simulacao_saude_publica_brasil.csv`) contém registros temporais e geográficos contendo as seguintes variáveis:
- **ano, mes, data:** Referência temporal dos registros.
- **regiao, uf, municipio:** Localização geográfica do município analisado.
- **expectativa_vida:** Média de expectativa de vida da população.
- **taxa_mortalidade:** Índice de mortalidade registrado.
- **taxa_internacao:** Taxa de internações hospitalares.
- **cobertura_vacinal:** Percentual de cobertura de imunização.
- **medicos_por_1000:** Número de médicos por mil habitantes.
- **leitos_hospitalares:** Quantidade total de leitos disponíveis.
- **casos_doencas_cronicas:** Incidência de casos crônicos.
- **nivel_criticidade:** Classificação do nível de criticidade sistêmica.

## 3. Leitura dos dados
Leitura inicial da base de dados utilizando a biblioteca Pandas.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Leitura da base de dados
df = pd.read_csv('../dados/simulacao_saude_publica_brasil.csv')
df.head()
```

## 4. Limpeza e preparação
Verificação e tratamento de valores nulos e inconsistências na base.

```python
# Verificando valores nulos
print(df.isnull().sum())
# Informações gerais dos tipos de dados
df.info()
```

## 5. Engenharia de atributos
Criação de novas colunas e métricas úteis para a análise.

```python
# Exemplo de engenharia de atributos
df['ano_periodo'] = df['ano'].astype(str) + '-' + df['mes'].astype(str).str.zfill(2)
df.head(2)
```

## 6. Análise exploratória
Estatísticas descritivas básicas dos principais indicadores.

```python
df.describe()
```

## 7. KPIs
Cálculo dos principais Indicadores Chave de Desempenho.

```python
media_expectativa = df['expectativa_vida'].mean()
media_mortalidade = df['taxa_mortalidade'].mean()
media_vacina = df['cobertura_vacinal'].mean()

print(f"Média Geral de Expectativa de Vida: {media_expectativa:.2f} anos")
print(f"Média Geral de Taxa de Mortalidade: {media_mortalidade:.2f}")
print(f"Média Geral de Cobertura Vacinal: {media_vacina:.2f}%")
```

## 8. Gráficos
Geração de visualizações gráficas para análise visual dos dados.

```python
plt.figure(figsize=(10, 5))
sns.barplot(data=df, x='regiao', y='expectativa_vida', color='#2e7d32', ci=None)
plt.title('Expectativa de Vida Média por Região')
plt.xlabel('Região')
plt.ylabel('Expectativa de Vida')
plt.show()
```

## 9. Interpretação dos resultados

A análise exploratória dos dados revela padrões fundamentais sobre a saúde pública brasileira entre 2015 e 2024, evidenciando que as regiões que registram maior expectativa de vida e menores taxas de mortalidade estão diretamente correlacionadas com índices elevados de cobertura vacinal e grande densidade de recursos médicos.

Através dos filtros interativos e da visualização gráfica, observa-se que desigualdades regionais ainda impactam a eficiência assistencial, demonstrando que investimentos contínuos em infraestrutura de atendimento e prevenção epidemiológica são determinantes cruciais para a mitigação de quadros críticos e para a melhoria geral da qualidade de vida populacional.
